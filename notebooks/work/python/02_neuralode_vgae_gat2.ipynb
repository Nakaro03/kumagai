### 2.2 発明者とトピックのグラフ構築の際のエッジの重みを計算

「発明者inventorが、技術トピックtopicに対してどれくらい深く関わっているか」
sim_score (特許のトピック適合度) × RoleWeight (筆頭/共同の重み) 

```python
df.head(5)
```

```python
import pandas as pd
import numpy as np
import torch
from torch_geometric.data import Data

# --- 設定 ---
SIM_SCORE_THRESHOLD = 0.5 
LEAD_WEIGHT = 1.0  # 筆頭発明者の重み係数
CO_WEIGHT = 0.5    # 共同発明者の重み係数

def preprocess_weighted_edges(df):

    print("Lead Inventor列を使用した重み付きデータ作成中...")
    
    # 必要な列をコピー (SettingWithCopyWarning回避)
    sub_df = df[['year_month', 'inventors', 'lead_inventor', 'topic_id', 'sim_score']].copy()
    
    # sim_scoreによる足切り
    sub_df = sub_df[sub_df['sim_score'] >= SIM_SCORE_THRESHOLD]
    
    # 発明者リストのパース関数
    def parse_inventors(x):
        if isinstance(x, list): return x
        if pd.isna(x) or x == "": return []
        # 文字列の場合、全角/半角カンマで分割
        return [i.strip() for i in str(x).replace('，', ',').split(',') if i.strip()]

    processed_rows = []
    
    for _, row in sub_df.iterrows():
        # 発明者リスト取得
        invs = parse_inventors(row['inventors'])
        if not invs:
            continue
            
        # 筆頭発明者名取得 (空白除去)
        lead_name = str(row['lead_inventor']).strip() if pd.notna(row['lead_inventor']) else ""
        
        for inv_name in invs:
            inv_name_clean = inv_name.strip()
            
            # 【重要】名前の一致判定
            is_lead = (inv_name_clean == lead_name)
            
            # 役割による重み係数
            role_w = LEAD_WEIGHT if is_lead else CO_WEIGHT
            
            # 最終的なエッジ重み = トピック類似度 * 役割重み
            # sim 0.9 * lead 1.0 = 0.9 （類似度 * 筆頭発明者）
            # sim 0.9 * co 0.5 = 0.45 （類似度 * 共同発明者)
            final_weight = row['sim_score'] * role_w
            
            processed_rows.append({
                'year_month': row['year_month'],
                'inventor': inv_name_clean,
                'topic_id': int(row['topic_id']),
                'weight': final_weight,
                'is_lead': is_lead # 確認用
            })
            
    new_df = pd.DataFrame(processed_rows)
    # 日付型変換
    new_df['year_month'] = pd.to_datetime(new_df['year_month'], format='%Y-%m-%d') # 必要に応じてフォーマット調整
    
    print(f"処理完了: エッジ総数 {len(new_df)}")
    print(f"うち筆頭発明者エッジ数: {new_df['is_lead'].sum()}")
    
    return new_df

weighted_edges_df = preprocess_weighted_edges(df)
print(weighted_edges_df.head())
```

### 2.3 重み付きグラフの構築

```python
# 1. ノードIDのマッピング作成
# 全データからユニークなリストを作成
all_inventors = sorted(list(set(weighted_edges_df['inventor'])))
all_topics = sorted(list(set(weighted_edges_df['topic_id'])))

num_inventors = len(all_inventors)
num_topics = len(all_topics)
total_nodes = num_inventors + num_topics

print(f"発明者数: {num_inventors}")
print(f"トピック数: {num_topics}")
print(f"総ノード数: {total_nodes}")
```

```python
# ID辞書の作成
# 発明者ID: 0 〜 (num_inventors - 1)
inventor_to_idx = {name: i for i, name in enumerate(all_inventors)}

# トピックID: num_inventors 〜 (total_nodes - 1)
topic_to_idx = {tid: i + num_inventors for i, tid in enumerate(all_topics)}

# 確認: 最初の5人のマッピングを表示
print("\n--- IDマッピング例 (発明者) ---")
print(list(inventor_to_idx.items())[:5])

print("\n--- IDマッピング例 (トピック) ---")
print(list(topic_to_idx.items())[:5])
```

```python
# 2. 年ごとにグループ化
year_groups = weighted_edges_df.groupby(weighted_edges_df['year_month'].dt.year)

# 確認: どんな年が含まれているか
available_years = sorted(year_groups.groups.keys())
print(f"データに含まれる年: {available_years}")

# テスト用に最初の年を取得して中身を見てみる
test_year = available_years[0]
test_group = year_groups.get_group(test_year)

print(f"\n--- {test_year}年のデータプレビュー ({len(test_group)}件) ---")
print(test_group[['inventor', 'topic_id', 'weight']].head())
```

```python
# 3. エッジの重みの集計  (test_year分のみ実行)
edge_weights_dict = {}
active_nodes = set()

print(f"--- {test_year}年の重み集計を開始 ---")

for _, row in test_group.iterrows():
    # IDに変換
    if row['inventor'] not in inventor_to_idx or row['topic_id'] not in topic_to_idx:
        continue

    u = inventor_to_idx[row['inventor']]
    v = topic_to_idx[row['topic_id']]
    w = row['weight']

    # ペアをソートしてタプル化 (小さいID, 大きいID)
    pair = tuple(sorted((u, v)))

    if pair not in edge_weights_dict:
        edge_weights_dict[pair] = 0.0

    # 重みを加算
    edge_weights_dict[pair] += w

    active_nodes.add(u)
    active_nodes.add(v)

# 集計結果の確認（最初の5件）
print(f"集計後のユニークなエッジ数: {len(edge_weights_dict)}")
print("エッジ重みのサンプル（ノードIDペア: 重み):")
for i, (pair, weight) in enumerate(edge_weights_dict.items()):
    if i>=5: break
    inv_id, top_id = pair 

    inv_name = all_inventors[inv_id] if inv_id < num_inventors else "Topic?"
    top_name = all_topics[top_id - num_inventors] if top_id >= num_inventors else "Inv?"

    print(f"  {pair} ({inv_name} - Topic {top_name}): {weight:.4f}")
```

```python
# 「特定の発明者」と「特定の技術トピック」のIDの組み合わせ
pair
```

```python
# 4. 全期間のグラフ構築を実行
graphs = {}

print("全期間のグラフ構築を開始...")

for year, group in year_groups:
    edges_weights_dict = {}
    active_nodes = set()

    for _, row in group.iterrows():
        u = inventor_to_idx[row['inventor']]
        v = topic_to_idx[row['topic_id']]
        w = row['weight']

        pair = tuple(sorted((u, v)))

        if pair not in edge_weights_dict:
            edge_weights_dict[pair] = 0.0
        edge_weights_dict[pair] += w

        active_nodes.add(u)
        active_nodes.add(v)

    if not edge_weights_dict:
        continue

    # COO形式 (edge_index) と edge_weight の作成
    edges = []
    weights = []

    for (u, v), w in edge_weights_dict.items():
        # 双方向 (u -> v)
        edges.append([u, v])
        weights.append(w)
        # 双方向 (v -> u)
        edges.append([v, u])
        weights.append(w)
        
    # Tensor変換
    edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()
    edge_weight = torch.tensor(weights, dtype=torch.float32)
    
    # アクティブなノードのマスク作成
    active_mask = torch.zeros(total_nodes, dtype=torch.bool)
    active_mask[list(active_nodes)] = True

    # Dataオブジェクト作成
    graphs[year] = Data(
        edge_index=edge_index,
        edge_weight=edge_weight,
        num_nodes=total_nodes,
        active_mask=active_mask,
        year=year
    )

print(f"完了: {len(graphs)}年分のグラフ辞書 (graph_dict) が完成しました。")
```

### 2.4 モデル定義 (重み対応型 GCN-VGAE)

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torchdiffeq import odeint

print('ライブラリのインポート')
```

```python
# 重み対応エンコーダ
class WeightedEncoder(nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super().__init__()
        # GCNConvは edge_weight を引数に取れます
        self.conv1 = GCNConv(in_channels, hidden_channels * 2)
        self.conv2 = GCNConv(hidden_channels * 2, hidden_channels)
        self.conv_mu = GCNConv(hidden_channels, out_channels)
        self.conv_logvar = GCNConv(hidden_channels, out_channels)
        self.dropout = nn.Dropout(0.2)
        self.batch_norm1 = nn.BatchNorm1d(hidden_channels * 2)
        self.batch_norm2 = nn.BatchNorm1d(hidden_channels)

    def forward(self, x, edge_index, edge_weight):
        # edge_weight を渡すことで、筆頭発明者のリンクを重視して畳み込み
        x = F.relu(self.batch_norm1(self.conv1(x, edge_index, edge_weight)))
        x = self.dropout(x)
        x = F.relu(self.batch_norm2(self.conv2(x, edge_index, edge_weight)))
        x = self.dropout(x)
        return self.conv_mu(x, edge_index, edge_weight), self.conv_logvar(x, edge_index, edge_weight) 
```

```python
# ODE予測器 (潜在空間での時間発展を学習)
class SimpleODEPredictor(nn.Module):
    def __init__(self, latent_dim, hidden_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim + 1, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, latent_dim)
        )
    def forward(self, t, z):
        # 時間 t をベクトル化して結合
        t_vec = t.expand(z.size(0), 1)
        return self.net(torch.cat([z, t_vec], dim=1))
```

```python
# 発明者-トピック統合モデル
class WeightedInventorTopicModel(nn.Module):
    def __init__(self, num_inventors, num_topics, embedding_dim=64, hidden_dim=32, latent_dim=16):
        super().__init__()
        self.num_inventors = num_inventors
        self.num_topics = num_topics

        # ノードIDに対するEmbedding
        self.inventor_emb = nn.Embedding(num_inventors, embedding_dim)
        self.topic_emb = nn.Embedding(num_topics, embedding_dim)
        nn.init.xavier_normal_(self.inventor_emb.weight)
        nn.init.xavier_normal_(self.topic_emb.weight)

        # 重み対応エンコーダ
        self.encoder = WeightedEncoder(embedding_dim, hidden_dim, latent_dim)
        
        # ODE
        self.ode_func = SimpleODEPredictor(latent_dim, hidden_dim)
        
        # リンク予測 (デコーダ)
        self.link_predictor = nn.Sequential(
            nn.Linear(latent_dim * 2, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1)
        )

    def get_node_features(self, device):
        inv_feats = self.inventor_emb(torch.arange(self.num_inventors, device=device))
        topic_feats = self.topic_emb(torch.arange(self.num_topics, device=device))
        return torch.cat([inv_feats, topic_feats], dim=0)

    def encode(self, edge_index, edge_weight):
        x = self.get_node_features(edge_index.device)
        mu, logvar = self.encoder(x, edge_index, edge_weight)
        
        if self.training:
            std = torch.exp(0.5 * logvar)
            z = mu + torch.randn_like(std) * std
        else:
            z = mu
        return z, mu, logvar

    def decode(self, z, edge_index):
        z_norm = F.normalize(z, p=2, dim=1)
        src, dst = edge_index
        dot = (z_norm[src] * z_norm[dst]).sum(dim=1)
        logits = dot * 10.0
        return torch.sigmoid(logits)
    
    def predict_future(self, z):
        # ODEで未来の状態を予測 (t=0 -> t=1)
        t = torch.tensor([0., 1.0]).to(z.device)
        # odeint は [z(t0), z(t1)] を返すので [-1] で z(t1) を取得
        z_next = odeint(self.ode_func, z, t, method='dopri5')[-1]
        return z_next
```

### 3.4 model学習実行

```python
import time
from torch_geometric.utils import negative_sampling

# デバイス設定
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# モデル初期化
model = WeightedInventorTopicModel(
    num_inventors=num_inv, 
    num_topics=num_top
).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-3)

print("重み付きモデルの学習開始...")
years = sorted(graph_dict.keys())
start_time = time.time()

# 学習ループ
for epoch in range(100):
    model.train()
    total_loss = 0
    
    for i in range(len(years) - 1):
        year_t = years[i]
        year_t1 = years[i+1]
        
        data_t = graph_dict[year_t].to(device)
        data_t1 = graph_dict[year_t1].to(device)
        
        optimizer.zero_grad()
        
        # 1. 重み付きグラフで現在の状態をエンコード
        z_t, mu, logvar = model.encode(data_t.edge_index, data_t.edge_weight)
        
        # 2. 現在のリンク再構成誤差
        pos_pred = model.decode(z_t, data_t.edge_index)
        pos_loss = -torch.log(pos_pred + 1e-15).mean()
        
        # 負例サンプリング
        neg_edge_index = negative_sampling(
            edge_index=data_t.edge_index,
            num_nodes=data_t.num_nodes,
            num_neg_samples=data_t.edge_index.size(1) # 正例と同じ数
        )
        neg_pred = model.decode(z_t, neg_edge_index)
        neg_loss = -torch.log(1 - neg_pred + 1e-15).mean()
        recon_loss = pos_loss + neg_loss

        # 3. KL Divergence
        kl_loss = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
        
        # 4. 未来予測
        z_next = model.predict_future(mu)
        
        # 5. 未来のリンク予測誤差 (t+1 のグラフ構造を予測)
        # --- 5. 未来のリンク予測学習 ---
        if data_t1.edge_index.numel() > 0:
            # (A) 未来の正例
            fut_pos_pred = model.decode(z_next, data_t1.edge_index)
            fut_pos_loss = -torch.log(fut_pos_pred + 1e-15).mean()
            
            # (B) 未来の負例（★ここも追加）
            fut_neg_edge_index = negative_sampling(
                edge_index=data_t1.edge_index,
                num_nodes=data_t1.num_nodes,
                num_neg_samples=data_t1.edge_index.size(1)
            )
            fut_neg_pred = model.decode(z_next, fut_neg_edge_index)
            fut_neg_loss = -torch.log(1 - fut_neg_pred + 1e-15).mean()
            
            future_loss = fut_pos_loss + fut_neg_loss
        else:
            future_loss = 0
            
        # 損失の合計
        loss = recon_loss + 0.0 * kl_loss + 10*future_loss
        
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        total_loss += loss.item()
        
    if (epoch+1) % 5 == 0:
        print(f"Epoch {epoch+1}: Total Loss {total_loss:.4f}")

print(f"学習完了 (所要時間: {time.time() - start_time:.2f}秒)")
```

### 3.5 評価と未来予測データの抽出

```python
import pandas as pd
import numpy as np
import torch

# --- 1. 必要な情報の準備 ---

# 最後の2期間（学習の最終ステップ）を取得
years = sorted(graph_dict.keys())
last_year_t = years[-2]
last_year_t1 = years[-1]

data_t = graph_dict[last_year_t].to(device)
data_t1 = graph_dict[last_year_t1].to(device) # 検証用データ（実際のリンク）

num_inventors = model.num_inventors
num_topics = model.num_topics
total_nodes = num_inventors + num_topics

print(f"最終学習期間: {last_year_t}")
print(f"予測対象期間: {last_year_t1}")
```

```python
# --- 2. 実際のリンクセット (正解データ) の作成 ---

actual_edges = data_t1.edge_index.t().cpu().numpy()

actual_links_set = set()
for u, v in actual_edges:
    if u < num_inventors and v >= num_inventors:
        actual_links_set.add((u, v))

print(f"正解リンク（Inventor->Topic, t+1期間）の数: {len(actual_links_set)}")
```

```python
# --- 3. 未来予測の実行 ---

model.eval()
with torch.no_grad():
    # t時点のグラフ構造から潜在ベクトル mu をエンコード
    _, mu_t, _ = model.encode(data_t.edge_index, data_t.edge_weight)
    
    # mu_t をODEに通し、t+1時点の潜在ベクトルを予測
    z_pred = model.predict_future(mu_t)

# マッピング情報の再構築 (セル2の処理に依存)
all_inventors = sorted(list(set(weighted_edges_df['inventor'])))
all_topics = sorted(list(set(weighted_edges_df['topic_id'])))
idx_to_inventor = {i: name for i, name in enumerate(all_inventors)}
idx_to_topic = {i + num_inventors: tid for i, tid in enumerate(all_topics)}
```

```python
# --- 4. 予測スコアの計算とデータフレーム化 ---

inventor_indices = torch.arange(num_inventors, device=device)
topic_indices = torch.arange(num_inventors, total_nodes, device=device)

predicted_links = []

# 発明者-トピックの総当たり戦
for inv_idx in inventor_indices:
    
    z_c = z_pred[inv_idx].unsqueeze(0).expand(num_topics, -1)
    z_p = z_pred[topic_indices]
    combined = torch.cat([z_c, z_p], dim=1)
    scores = model.link_predictor(combined).squeeze()
    
    probabilities = torch.sigmoid(scores).detach().cpu().numpy()
    
    inventor_name = idx_to_inventor[inv_idx.item()]
    
    for i, prob in enumerate(probabilities):
        topic_global_idx = topic_indices[i].item()
        topic_id = idx_to_topic[topic_global_idx]
        
        is_correct = (inv_idx.item(), topic_global_idx) in actual_links_set

        predicted_links.append({
            'inventor': inventor_name,
            'topic_id': topic_id,
            'predicted_score': prob,
            'is_actual_link_in_t+1': is_correct # 正解フラグ
        })
```

```python
# --- 5. データフレーム化と表示 ---

prediction_df = pd.DataFrame(predicted_links)
print("\n--- 予測結果のデータフレーム（上位10件） ---")

# スコアの高い順にソートし、正解フラグを確認
top_predictions = prediction_df.sort_values(by='predicted_score', ascending=False)
print(top_predictions)
```

```python
# 文字列 'True' ではなく、ブール値の True を使用します
correct_predictions = prediction_df[prediction_df['is_actual_link_in_t+1'] == True]

print("--- 実際に t+1 期間でリンクが形成された（正解）データ ---")

# スコアの高い順にソートして表示
# 正解データの中でも、モデルが高いスコアを出したものを確認すると、モデルの学習状況がわかります。
print(correct_predictions.sort_values(by='predicted_score', ascending=False).head(10))

# 正解データの総数
print(f"\n抽出された正解リンクの総数: {len(correct_predictions)}")
```

```python
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score

y_true = prediction_df['is_actual_link_in_t+1'].values
y_scores = prediction_df['predicted_score'].values

# 1. AUC Score (全体でのランキングの総合性能)
# 1.0に近いほど良い。これが「全体での正解率」の最も適切な指標です。
auc_score = roc_auc_score(y_true, y_scores)
print(f"1. AUC Score (全体でのランキング性能): {auc_score:.4f}")

# 2. Average Precision (AP)
# 正例の少ないデータセットでの実用的な予測性能。1.0に近いほど良い。
ap_score = average_precision_score(y_true, y_scores)
print(f"2. Average Precision (AP): {ap_score:.4f}")
```

### データが悪いのかモデルが悪いのか？

```python
from sklearn.metrics import roc_auc_score
import torch.nn.functional as F

model.eval()
with torch.no_grad():
    # 最新の学習データ（例: 最後の年）
    year = years[-1]
    data = graph_dict[year].to(device)
    
    # 1. エンコード
    z, _, _ = model.encode(data.edge_index, data.edge_weight)
    
    # 2. 「現在の」リンクを正しくスコアリングできるかチェック
    # 正例スコア
    pos_score = model.decode(z, data.edge_index).cpu().numpy()
    
    # 負例スコア（正例と同じ数だけランダム生成）
    neg_edge_index = negative_sampling(data.edge_index, num_nodes=data.num_nodes, num_neg_samples=len(pos_score))
    neg_score = model.decode(z, neg_edge_index).cpu().numpy()
    
    # AUC計算
    y_true = [1] * len(pos_score) + [0] * len(neg_score)
    y_scores = np.concatenate([pos_score, neg_score])
    
    recon_auc = roc_auc_score(y_true, y_scores)

    if recon_auc >= 0.8:
        print(f"Reconstruction AUC (現状復元能力): {recon_auc:.4f} GCN（エンコーダ）は正常。悪いのは「未来予測（predict_future）」の部分")
    elif recon_auc < 0.6 and recon_auc >= 0.4:
        print(f"Reconstruction AUC (現状復元能力): {recon_auc:.4f} データセットの作り方（IDズレなど）か、GCNの構造自体に問題あり。")
```
