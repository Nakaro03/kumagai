#### 3.1 メタデータ埋め込みの取得と結合
特許番号で各埋め込み辞書からベクトルを取得（なければ0埋め）テキスト埋め込みと連結（metadata_embedding）

```python
import pickle
import pandas as pd
import numpy as np

# 1. メタデータ辞書の読み込み
EMBEDS_DIR_PATH = "/home/nakamuraroi/kumagai/work/embeds/"
with open(f"{EMBEDS_DIR_PATH}patent_embeds_dic.pkl", "rb") as f:
    patent_embeds_dic = pickle.load(f)

print(f"Loaded dictionary with {len(patent_embeds_dic)} entries.")

# 2. マッピング処理
key_column = 'patent_number' 

if key_column not in kg_dataset.columns:
    print(f"Error: '{key_column}' カラムが見つかりません。kg_datasetのカラム名を確認してください。")
else:
    # map関数を使って辞書からベクトルを取得し、新しい列に代入
    kg_dataset['metadata_embedding'] = kg_dataset[key_column].map(patent_embeds_dic)

    # 3. 欠損値（辞書にキーがなかった特許）の確認と処理
    missing_count = kg_dataset['metadata_embedding'].isna().sum()
    
    if missing_count > 0:
        print(f"注意: {missing_count} 件の特許についてメタデータ埋め込みが見つかりませんでした。")
        
        # 必要であれば、欠損を「ゼロベクトル」で埋める処理（オプション）
        sample_vec = next(iter(patent_embeds_dic.values()))
        zero_vec = np.zeros_like(sample_vec)
        
        # NaN の箇所をゼロベクトルで埋める
        kg_dataset['metadata_embedding'] = kg_dataset['metadata_embedding'].apply(
            lambda x: x if isinstance(x, (list, np.ndarray)) else zero_vec
        )
        print("欠損値をゼロベクトルで補完しました。")

    # --- 結果確認 ---
    print("\nColumns added. Check first 3 rows:")
    print(kg_dataset[[key_column, 'description_embedding', 'metadata_embedding']].head(3))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def visualize_individual_vectors(df, max_display=500, title="Individual Inventor Trajectories"):
    """
    各発明者の個別の移動ベクトル(x, y) -> (x+dx, y+dy) をプロットする関数
    
    Args:
        df: 'x', 'y', 'dx', 'dy' カラムを持つDataFrame
        max_display: 表示する最大人数（多すぎると見えなくなるためサンプリングする）
        title: 図のタイトル
    """
    
    # データ数が多すぎる場合はランダムサンプリング
    if len(df) > max_display:
        print(f"Data count ({len(df)}) exceeds limit. Sampling {max_display} points for visualization.")
        df_plot = df.sample(n=max_display, random_state=42).copy()
    else:
        df_plot = df.copy()

    # ベクトルの方向（角度）によって色を変えるための計算
    # arctan2で -pi ～ pi の値を取得
    angles = np.arctan2(df_plot['dy'], df_plot['dx'])
    
    # 描画
    plt.figure(figsize=(12, 10))
    
    # 1. 開始点（現在の位置）を点でプロット
    plt.scatter(df_plot['x'], df_plot['y'], c='gray', s=10, alpha=0.5, label='Start Position')

    # 2. 移動ベクトルを矢印でプロット
    # angles='xy', scale_units='xy', scale=1 とすることで、
    # 実際のデータ座標系での (dx, dy) の長さそのままで描画されます。
    Q = plt.quiver(
        df_plot['x'], df_plot['y'], 
        df_plot['dx'], df_plot['dy'],
        angles,           # 色付けの基準（角度）
        cmap='hsv',       # 角度によって色が変わるカラーマップ (虹色)
        angles='xy', scale_units='xy', scale=1, 
        width=0.003, headwidth=4, headlength=5, alpha=0.8
    )
    
    # カラーバー（方向を示す）
    cbar = plt.colorbar(Q, label='Movement Direction (Angle)')
    cbar.set_ticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
    cbar.set_ticklabels(['-π (Left)', '-π/2 (Down)', '0 (Right)', 'π/2 (Up)', 'π (Left)'])

    # 凡例用の矢印（大きさの基準）
    # データ内の平均的な移動距離を基準キーとして表示
    avg_len = np.mean(np.sqrt(df_plot['dx']**2 + df_plot['dy']**2))
    plt.quiverkey(Q, 0.9, 1.02, avg_len, f'Avg Move: {avg_len:.2f}', labelpos='E', coordinates='axes')

    plt.title(title)
    plt.xlabel("Latent Dimension 1")
    plt.ylabel("Latent Dimension 2")
    plt.grid(True, linestyle=':', alpha=0.4)
    plt.axis('equal') # アスペクト比を揃えないと矢印の向きが歪んで見えるため重要
    plt.legend(loc='upper left')
    
    plt.show()

# ==========================================
# 実行例 (ダミーデータ)
# ==========================================
if __name__ == "__main__":
    np.random.seed(42)
    n_samples = 3000 
    
    # 2つのクラスタ
    x = np.concatenate([np.random.normal(3, 1.5, n_samples//2), np.random.normal(8, 1.5, n_samples//2)])
    y = np.concatenate([np.random.normal(3, 1.5, n_samples//2), np.random.normal(8, 1.5, n_samples//2)])
    
    # 移動ベクトル: クラスタ中心から外へ広がるような動き + ランダムノイズ
    # (3,3)付近のグループは右上に、(8,8)付近は左下に動くなど
    dx = (np.random.rand(n_samples) - 0.5) * 2
    dy = (np.random.rand(n_samples) - 0.5) * 2
    
    # 少し構造を持たせる（例: 全体的に時計回りの渦っぽく）
    # dx += -(y - 5.5) * 0.2
    # dy += (x - 5.5) * 0.2

    df_inventors = pd.DataFrame({'x': x, 'y': y, 'dx': dx, 'dy': dy})

    # 生データ（個別のベクトル）を表示
    # max_display=300 くらいに絞ると個々の動きが見やすいです
    visualize_individual_vectors(df_inventors, max_display=300)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binned_statistic_2d

def visualize_grid_averaged_flow(df, grid_size=20, min_count=3, title="Grid Averaged Vector Field"):
    """
    空間をグリッドに切り、各セル内のベクトルの平均値を表示する関数
    
    Args:
        df: 'x', 'y', 'dx', 'dy' を持つDataFrame
        grid_size: グリッドの分割数（大きいほど細かくなる）
        min_count: 平均算出に必要な最小データ数（これ未満のセルの矢印は表示しない）
    """
    
    # 1. 範囲決定
    x_min, x_max = df['x'].min() - 0.5, df['x'].max() + 0.5
    y_min, y_max = df['y'].min() - 0.5, df['y'].max() + 0.5
    
    # 2. グリッドごとの統計量を計算
    # dxの平均
    mean_dx, x_edge, y_edge, _ = binned_statistic_2d(
        df['x'], df['y'], df['dx'], 
        statistic='mean', bins=grid_size, range=[[x_min, x_max], [y_min, y_max]]
    )
    # dyの平均
    mean_dy, _, _, _ = binned_statistic_2d(
        df['x'], df['y'], df['dy'], 
        statistic='mean', bins=grid_size, range=[[x_min, x_max], [y_min, y_max]]
    )
    # データ数（信頼性確保のため）
    count, _, _, _ = binned_statistic_2d(
        df['x'], df['y'], df['dx'], 
        statistic='count', bins=grid_size, range=[[x_min, x_max], [y_min, y_max]]
    )

    # 3. グリッド中心座標の作成
    x_center = (x_edge[:-1] + x_edge[1:]) / 2
    y_center = (y_edge[:-1] + y_edge[1:]) / 2
    X, Y = np.meshgrid(x_center, y_center)
    
    # 転置（binned_statistic_2dの戻り値の形状合わせ）
    mean_dx = mean_dx.T
    mean_dy = mean_dy.T
    count = count.T

    # 4. ノイズ除去（データが少なすぎるグリッドは表示しない）
    mask = count >= min_count
    
    # 5. ベクトルの強さ（速さ）を計算（色付け用）
    speed = np.sqrt(mean_dx**2 + mean_dy**2)
    
    # --- 描画 ---
    plt.figure(figsize=(12, 10))
    
    # 背景：元のデータ点を薄く表示（分布の参考用）
    plt.scatter(df['x'], df['y'], c='lightgray', s=5, alpha=0.2, label='Raw Data Points')

    # 前景：平均ベクトル場
    # 色を速度(speed)に対応させ、プラズマカラーなどで表示
    Q = plt.quiver(
        X[mask], Y[mask], 
        mean_dx[mask], mean_dy[mask],
        speed[mask],    # 色付けの基準
        cmap='plasma',  # 速度が速いほど明るい色に
        angles='xy', scale_units='xy', scale=1, # 生データと同じスケールで表示
        width=0.003, headwidth=4, headlength=5
    )
    
    plt.colorbar(Q, label='Avg Speed (Magnitude)')
    
    # 凡例用の矢印
    # 表示されている矢印の平均的な長さを凡例にする
    if np.any(mask):
        avg_len = np.mean(speed[mask])
        plt.quiverkey(Q, 0.9, 1.02, avg_len, f'Avg Speed: {avg_len:.2f}', labelpos='E', coordinates='axes')

    plt.title(f"{title} (Grid: {grid_size}x{grid_size}, Min Count: {min_count})")
    plt.xlabel("Latent Dimension 1")
    plt.ylabel("Latent Dimension 2")
    plt.grid(True, linestyle=':', alpha=0.4)
    plt.legend(loc='upper left')
    
    plt.show()

# ==========================================
# 実行例
# ==========================================
if __name__ == "__main__":
    np.random.seed(42)
    n_samples = 3000
    
    # データを生成
    x = np.concatenate([np.random.normal(3, 1.5, n_samples//2), np.random.normal(8, 1.5, n_samples//2)])
    y = np.concatenate([np.random.normal(3, 1.5, n_samples//2), np.random.normal(8, 1.5, n_samples//2)])
    dx = (np.random.rand(n_samples) - 0.5) * 2
    dy = (np.random.rand(n_samples) - 0.5) * 2
    
    # 少しトレンドを作る（中心に向かう動きなど）
    dx += (5.5 - x) * 0.1
    dy += (5.5 - y) * 0.1
    
    df_inventors = pd.DataFrame({'x': x, 'y': y, 'dx': dx, 'dy': dy})

    # グリッド平均化して表示
    # grid_size=20 程度が全体像を見るのに適しています
    visualize_grid_averaged_flow(df_inventors, grid_size=20, min_count=5)
```
